```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import json
```

```python
# Load Evaluation Data
# Assuming you save your evaluation results to a JSON file like this
# The structure is a list of dictionaries, one for each model run.
data = [
    {'model_name': 'TR-MODEL-1', 'bert_score': 0.870, 'bleu': 0.082, 'rougeL': 0.225, 'chrf++': 0.208},
    {'model_name': 'TR-MODEL-2', 'bert_score': 0.868, 'bleu': 0.046, 'rougeL': 0.234, 'chrf++': 0.563},
    {'model_name': 'TR-MODEL-3', 'bert_score': 0.869, 'bleu': 0.045, 'rougeL': 0.232, 'chrf++': 0.564},
    {'model_name': 'LIN-MODEL-1', 'bert_score': 0.852, 'bleu': 0.006, 'rougeL': 0.169, 'chrf++': 0.382},
    # Add data from your other runs here
]
```

```python
df = pd.DataFrame(data)
df.set_index('model_name', inplace=True)
```

```python
# Plot Heatmap of All Models
plt.figure(figsize=(10, 6))
sns.heatmap(df, annot=True, fmt=".3f", cmap="viridis")
plt.title("Comparative Analysis of Model Performance")
plt.show()
```

```python
# Plot Bar Chart for a Specific Metric (e.g., BERTScore)
plt.figure(figsize=(10, 6))
df['bert_score'].sort_values().plot(kind='bar', color=sns.color_palette('viridis', len(df)))
plt.title("BERTScore Comparison Across Models")
plt.ylabel("BERTScore F1")
plt.xticks(rotation=45, ha='right')
plt.show()
```
